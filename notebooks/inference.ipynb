# DOT Inference Notebook

This notebook generates `predictions.csv` from saved model artifacts.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if (ROOT / "src").exists():
    sys.path.append(str(ROOT / "src"))
else:
    sys.path.append(str(ROOT.parent / "src"))

from dot.infer import build_arg_parser, infer
from dot.validate import validate_predictions
from dot.config import TEST_PATH
```

```python
args = build_arg_parser().parse_args([])
args.output_path = Path("predictions.csv")
infer(args)
validate_predictions(Path("predictions.csv"), TEST_PATH)
print("Inference completed successfully.")
```

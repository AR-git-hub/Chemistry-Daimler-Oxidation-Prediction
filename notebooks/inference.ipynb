# Inference Notebook (финальная версия под ТЗ)

Этот ноутбук оставляет только то, что нужно для сдачи:
1. Показать, как получить `predictions.csv` из уже обученной модели (inference).
2. Проверить формат итогового файла.

Важно:
- Обучение модели выполняется отдельным кодом: `scripts/train.py`.
- Получение предсказаний выполняется отдельным кодом: `scripts/predict.py`.
- Здесь мы запускаем inference-процедуру и формируем итоговый `predictions.csv`.

```python
# Режим выгрузки

Выбери в следующей ячейке `MODE`:
- `historical_best` — файл с лучшим подтвержденным LB.
- `final_candidate` — новый финальный кандидат на последний сабмит.
```

```python
from pathlib import Path
import shutil
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "scripts").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

sys.path.append(str(PROJECT_ROOT / "src"))
from dot.validate import validate_predictions

# Режим выбора итогового файла:
MODE = "final_candidate"  # historical_best | final_candidate

FILES_BY_MODE = {
    "historical_best": "predictions_submit_blend_97wave2_03sumpoolstack.csv",
    "final_candidate": "predictions_submit_blend_965wave2_03sumpool_005v18ens4.csv",
}

source_name = FILES_BY_MODE[MODE]
src = PROJECT_ROOT / source_name
dst = PROJECT_ROOT / "predictions.csv"

assert src.exists(), f"Файл не найден: {src}"
shutil.copyfile(src, dst)

validate_predictions(dst, PROJECT_ROOT / "data/merged/test_full.csv")

pred = pd.read_csv(dst)
print("MODE:", MODE)
print("Скопировано:", src.name)
print("Итог:", dst)
print("Rows:", len(pred), "| Cols:", list(pred.columns))
pred.head(3)
```
